# Employee Attrition - Data Visualization

## Project Overview

This project focuses on visualizing employee attrition data using Python,
Matplotlib and Seaborn.

The objective is to create clear and meaningful visualizations and identify
important patterns related to employee attrition.

## 1. Importing Libraries

The following Python libraries are used for data analysis and visualization.

- Pandas: for data manipulation and analysis.
- NumPy: for numerical operations.
- Matplotlib: for creating visualizations.
- Seaborn: for creating statistical graphics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
##loading the Dataset

df = pd.read_csv("../data/WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

## 2. Monthly Income Distribution by Job Role

This visualization shows the distribution of monthly income across different job roles.

A boxplot is used to compare the income distributions and identify differences in salary levels between job roles.

It also provides information about the median, variability, and possible outliers in monthly income.

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='JobRole',
    y='MonthlyIncome'
)

plt.xticks(rotation=45, ha='right')
plt.title('Monthly Income Distribution by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Monthly Income')

plt.tight_layout()

plt.savefig('../visualizations_task3/monthly_income_by_jobrole.png')

plt.show()
```

### Interpretation

The visualization shows that monthly income varies considerably between job roles. Some job roles have higher median incomes, while others show a lower income range.

The boxplot also highlights the variability of income within each job role.

## 3. Age vs Monthly Income

This scatterplot explores the relationship between employee age and monthly income.

The employees are distinguished according to their attrition status to provide an additional perspective on the data.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Age',
    y='MonthlyIncome',
    hue='Attrition'
)

plt.title('Age vs Monthly Income')
plt.xlabel('Age')
plt.ylabel('Monthly Income')

plt.tight_layout()

plt.savefig('../visualizations_task3/age_vs_monthly_income.png')

plt.show()
```

### Interpretation

The scatterplot shows the distribution of monthly income across different age groups.

It can be observed that monthly income generally varies among employees of different ages. The visualization also shows the distribution of employees according to their attrition status.

## 4. Job Satisfaction Distribution

This visualization presents the distribution of employees according to their job satisfaction level.

A countplot is used to compare the number of employees in each satisfaction category.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='JobSatisfaction'
)

plt.title('Job Satisfaction Distribution')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Number of Employees')

plt.tight_layout()

plt.savefig('../visualizations_task3/job_satisfaction_distribution.png')

plt.show()
```

### Interpretation

The visualization shows how employees are distributed across the different job satisfaction levels.

The number of employees varies between the satisfaction categories, providing an overview of the overall satisfaction distribution in the dataset.

## 5. Distribution of Years at Company

This histogram shows the distribution of employees according to the number of years they have spent at the company.

A histogram is used to identify the most common ranges of employee tenure and observe the overall distribution.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='YearsAtCompany',
    bins=20,
    kde=True
)

plt.title('Distribution of Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Number of Employees')

plt.tight_layout()

plt.savefig('../visualizations_task3/years_at_company_distribution.png')

plt.show()
```

### Interpretation

The histogram shows that the number of years employees have spent at the company is concentrated in the lower ranges, while fewer employees have very long periods of service.

This provides an overview of employee tenure within the organization.

## 6. Employees by Education Field

This visualization presents the number of employees in each education field.

A horizontal countplot is used to make the comparison between the different education fields easier to read.

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y='EducationField'
)

plt.title('Employees by Education Field')
plt.xlabel('Number of Employees')
plt.ylabel('Education Field')

plt.tight_layout()

plt.savefig('../visualizations_task3/education_field_distribution.png')

plt.show()
```

### Interpretation

The visualization shows the distribution of employees across different education fields.

Some education fields are more represented in the dataset than others, which highlights the diversity of employees' educational backgrounds.
